```python
import numpy as np
```

```python
a = np.array([[1,2],
        [2,3]])

print(a)
print(a.shape)
```

```python
b = np.random.random([3,4])
c = np.random.rand(3,4)

print(b)
print(b.shape)

print(c)
print(c.shape)
```

```python
ones = np.ones((10,5))
zeros = np.zeros((2,2))

print("Ones:\n", ones)
print("Zeros:\n", zeros)
```

```python
a = np.random.rand(6,3)
a_transposed = a.T

print(a)
print("\n" + "-" * 30 + "\n")
print(a_transposed)
```

```python
n, p, m = 2, 3, 4

a = np.random.rand(n, p)
b = np.random.rand(p, m)

c = np.dot(a, b)

print(a)
print(a.shape)

print("\n")

print(b)
print(b.shape)

print("\n")

print(c)
print(c.shape)
```
